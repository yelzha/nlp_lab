```python
import os
import shutil

def add_postfix_to_files(parent_folder, subfolder_name, postfix):
    # Create the full path to the subfolder
    target_folder = os.path.join(parent_folder, subfolder_name)

    if not os.path.isdir(target_folder):
        print(f"The folder '{target_folder}' does not exist.")
        return

    for filename in os.listdir(target_folder):
        file_path = os.path.join(target_folder, filename)

        if os.path.isfile(file_path):
            name, ext = os.path.splitext(filename)
            new_name = f"{name}_{postfix}{ext}"
            new_path = os.path.join(target_folder, new_name)
            os.rename(file_path, new_path)
            print(f"Renamed: {filename} -> {new_name}")

def copy_files_to_new_path(parent_folder, subfolder_name, destination_path):
    # Build the path to the source subfolder
    source_folder = os.path.join(parent_folder, subfolder_name)

    # Check if source folder exists
    if not os.path.isdir(source_folder):
        print(f"Source folder '{source_folder}' does not exist.")
        return

    # Create destination path if it doesn't exist
    os.makedirs(destination_path, exist_ok=True)

    # Loop through all files in source folder
    for filename in os.listdir(source_folder):
        source_file = os.path.join(source_folder, filename)

        # Copy only files (not subdirectories)
        if os.path.isfile(source_file):
            destination_file = os.path.join(destination_path, filename)
            shutil.copy2(source_file, destination_file)
            print(f"Copied: {filename} → {destination_path}")
            
            
from glob import glob

for folder_path in glob("dataset/modified_dataset/*"):
    dtype = folder_path.split("\\")[-1]
    main_folder = "dataset/modified_dataset"
    parent_folder = f"{main_folder}/{dtype}"
    postfix = f"{dtype}"

    add_postfix_to_files(parent_folder, "gsm_dataset", postfix)
    add_postfix_to_files(parent_folder, "math_dataset", postfix)
    
    copy_files_to_new_path(parent_folder, "gsm_dataset", "gsm_dataset")
    copy_files_to_new_path(parent_folder, "math_dataset", "math_dataset")
```

```python
import json
import pandas as pd
import re

with open("r2-ata/gsm8k-r2ata.jsonl", "r", encoding="utf8") as f:
    data = json.load(f)

raw_text = data["text"]
fixed_text = "[" + raw_text.strip().rstrip(",") + "]"
entries = json.loads(fixed_text)

def extract_parts(entry):
    full = entry["question"]
    final_answer = entry["answer"]
    
    match = re.match(r"Question: (.*?)\\nAnswer: (.*)", full)
    if match:
        question = match.group(1)
        explanation = match.group(2)
    else:
        question = full
        explanation = ""
    
    return {
        "Question": question.strip(),
        "Answer": explanation.strip(),
        "answer": final_answer.strip()
    }

processed = [extract_parts(e) for e in entries]

df = pd.DataFrame(processed)
df
```

```python
# AEDA method
def aeda(sentence, punctuations=None, ratio=0.3):
    if punctuations is None:
        punctuations = ['.', ',', '!', '?', ';', ':']
    words = sentence.split()
    n_insertions = max(1, int(len(words) * ratio))

    for _ in range(n_insertions):
        insert_punc = random.choice(punctuations)
        insert_pos = random.randint(0, len(words))
        words.insert(insert_pos, insert_punc)

    return ' '.join(words)

# Apply AEDA to each sentence
df['aeda_01'] = df['question'].apply(lambda x: aeda(x, ratio=0.1))
df["aeda_01_length"] = df["aeda_01"].apply(lambda x: len(x.split()))

df['aeda_03'] = df['question'].apply(lambda x: aeda(x, ratio=0.3))
df["aeda_03_length"] = df["aeda_03"].apply(lambda x: len(x.split()))

df['aeda_05'] = df['question'].apply(lambda x: aeda(x, ratio=0.5))
df["aeda_05_length"] = df["aeda_05"].apply(lambda x: len(x.split()))

df[['question', 'aeda_01', 'aeda_03', 'aeda_05']]
```
